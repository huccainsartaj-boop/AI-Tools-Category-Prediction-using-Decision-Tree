```python
# ================================
# AI Tools Catalog 2026
# Load Dataset & Exploratory Data Analysis (EDA)
# ================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)

# Load Dataset
file_path = "/kaggle/input/datasets/sahityabiswas5/trending-ai-tools-catalog-with-github-stats-2026/ai_tools_catalog_2026.csv"

df = pd.read_csv(file_path)

# ================================
# Basic Information
# ================================

print("="*60)
print("First 5 Rows")
print("="*60)
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary (Numerical Columns):")
display(df.describe())

print("\nStatistical Summary (Categorical Columns):")
display(df.describe(include="object"))

# ================================
# Missing Values
# ================================

print("\nMissing Values:")
missing = df.isnull().sum().sort_values(ascending=False)
display(missing[missing > 0])

# ================================
# Duplicate Values
# ================================

print("\nDuplicate Rows:", df.duplicated().sum())

# ================================
# Data Types
# ================================

print("\nData Types:")
display(df.dtypes)

# ================================
# Unique Values
# ================================

print("\nUnique Values in Each Column:")
display(df.nunique())

# ================================
# Category Distribution
# ================================

if "Category" in df.columns:
    plt.figure(figsize=(10,5))
    sns.countplot(data=df, x="Category", order=df["Category"].value_counts().index)
    plt.xticks(rotation=45)
    plt.title("Category Distribution")
    plt.show()

# ================================
# Subcategory Distribution
# ================================

if "Subcategory" in df.columns:
    plt.figure(figsize=(12,5))
    sns.countplot(data=df, x="Subcategory",
                  order=df["Subcategory"].value_counts().head(15).index)
    plt.xticks(rotation=90)
    plt.title("Top 15 Subcategories")
    plt.show()

# ================================
# Open Source Distribution
# ================================

if "Is Open Source" in df.columns:
    plt.figure(figsize=(5,5))
    df["Is Open Source"].value_counts().plot.pie(
        autopct="%1.1f%%", startangle=90)
    plt.ylabel("")
    plt.title("Open Source Distribution")
    plt.show()

# ================================
# Domain Extension Distribution
# ================================

if "Domain Extension" in df.columns:
    plt.figure(figsize=(8,5))
    sns.countplot(data=df,
                  x="Domain Extension",
                  order=df["Domain Extension"].value_counts().head(10).index)
    plt.xticks(rotation=45)
    plt.title("Top Domain Extensions")
    plt.show()

# ================================
# GitHub Stars Distribution
# ================================

if "GitHub Stars" in df.columns:
    plt.figure(figsize=(8,5))
    sns.histplot(df["GitHub Stars"], bins=30, kde=True)
    plt.title("GitHub Stars Distribution")
    plt.show()

# ================================
# GitHub Forks Distribution
# ================================

if "GitHub Forks" in df.columns:
    plt.figure(figsize=(8,5))
    sns.histplot(df["GitHub Forks"], bins=30, kde=True)
    plt.title("GitHub Forks Distribution")
    plt.show()

# ================================
# Description Word Count
# ================================

if "Description Word Count" in df.columns:
    plt.figure(figsize=(8,5))
    sns.histplot(df["Description Word Count"], bins=20, kde=True)
    plt.title("Description Word Count Distribution")
    plt.show()

# ================================
# Correlation Matrix
# ================================

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10,6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()

# ================================
# Top AI Tool Categories
# ================================

if "Category" in df.columns:
    display(df["Category"].value_counts())

# ================================
# Top Subcategories
# ================================

if "Subcategory" in df.columns:
    display(df["Subcategory"].value_counts().head(15))

# ================================
# Top Domain Extensions
# ================================

if "Domain Extension" in df.columns:
    display(df["Domain Extension"].value_counts())

# ================================
# Top Programming Languages
# ================================

if "Primary Language" in df.columns:
    display(df["Primary Language"].value_counts())

print("\nEDA Completed Successfully!")
```

```python
# ==========================================
# Preprocessing (Single Cell)
# Target = Category
# ==========================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Copy dataset
data = df.copy()

# -----------------------------
# Remove unnecessary columns
# -----------------------------
drop_cols = ["Name", "Website URL", "Domain", "Description"]

data = data.drop(columns=[col for col in drop_cols if col in data.columns])

# -----------------------------
# Handle Missing Values
# -----------------------------

# Numerical columns
num_cols = data.select_dtypes(include=np.number).columns
for col in num_cols:
    data[col] = data[col].fillna(data[col].median())

# Categorical columns
cat_cols = data.select_dtypes(include="object").columns
for col in cat_cols:
    data[col] = data[col].fillna(data[col].mode()[0])

# -----------------------------
# Convert Boolean Columns
# -----------------------------
bool_cols = data.select_dtypes(include="bool").columns
for col in bool_cols:
    data[col] = data[col].astype(int)

# -----------------------------
# Label Encoding
# -----------------------------
le = LabelEncoder()

for col in data.select_dtypes(include="object").columns:
    data[col] = le.fit_transform(data[col].astype(str))

# -----------------------------
# Features & Target
# -----------------------------
target = "Category"

X = data.drop(columns=[target])
y = data[target]

# -----------------------------
# Check Preprocessed Data
# -----------------------------
print("=" * 60)
print("Preprocessing Completed Successfully")
print("=" * 60)

print("Dataset Shape :", data.shape)
print("Feature Shape :", X.shape)
print("Target Shape  :", y.shape)

print("\nMissing Values:")
print(data.isnull().sum())

print("\nData Types:")
display(data.dtypes)

print("\nFirst 5 Rows:")
display(data.head())

print("\nTarget Classes:")
display(y.value_counts())
```

```python
# ==========================================
# Decision Tree Classifier (Single Cell)
# Removes rare classes automatically
# ==========================================

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Remove classes that have only one sample
class_counts = y.value_counts()
valid_classes = class_counts[class_counts >= 2].index

X_filtered = X[y.isin(valid_classes)]
y_filtered = y[y.isin(valid_classes)]

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_filtered,
    y_filtered,
    test_size=0.20,
    random_state=42,
    stratify=y_filtered
)

# Create Decision Tree Model
model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=10,
    random_state=42
)

# Train Model
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("="*50)
print("Decision Tree Classifier")
print("="*50)
print("Original Dataset Shape :", X.shape)
print("Filtered Dataset Shape :", X_filtered.shape)
print("Training Samples       :", X_train.shape[0])
print("Testing Samples        :", X_test.shape[0])
print("Accuracy               :", round(accuracy*100,2), "%")
```

```python
# ==========================================
# Decision Tree Evaluation & Visualization
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
from sklearn.tree import plot_tree

# ==========================
# Evaluation Metrics
# ==========================

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

print("="*60)
print("Decision Tree Evaluation")
print("="*60)

print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1 Score : {f1:.4f}")

print("\nClassification Report")
print("="*60)
print(classification_report(y_test, y_pred, zero_division=0))

# ==========================
# Confusion Matrix
# ==========================

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10,8))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    linewidths=0.5
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.show()

# ==========================
# Feature Importance
# ==========================

importance = model.feature_importances_

feature_importance = (
    pd.DataFrame({
        "Feature": X_filtered.columns,
        "Importance": importance
    })
    .sort_values(by="Importance", ascending=False)
)

print("\nFeature Importance")
display(feature_importance)

plt.figure(figsize=(10,6))
sns.barplot(
    data=feature_importance,
    x="Importance",
    y="Feature"
)
plt.title("Decision Tree Feature Importance")
plt.show()

# ==========================
# Decision Tree Visualization
# ==========================

plt.figure(figsize=(20,10))

plot_tree(
    model,
    feature_names=X_filtered.columns,
    filled=True,
    rounded=True,
    fontsize=8,
    max_depth=3
)

plt.title("Decision Tree (First 3 Levels)")
plt.show()
```
